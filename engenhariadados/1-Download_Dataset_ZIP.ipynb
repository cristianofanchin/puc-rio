## 📦 1.1. Baixar e extrair o ZIP contendo os dados em múltiplos arquivos CSV

```
import requests
import zipfile
import io
import os

extract_path = "/tmp/zip_data"
os.makedirs(extract_path, exist_ok=True)

# URL do Dataset no portal Maven Analytics
url_dataset = "https://maven-datasets.s3.amazonaws.com/Hospital+Patient+Records/Hospital+Patient+Records.zip"  

# Baixar ZIP
response = requests.get(url_dataset)
zip_bytes = io.BytesIO(response.content)

# Extrair apenas arquivos CSV válidos
with zipfile.ZipFile(zip_bytes, 'r') as z:
    for file in z.namelist():
        if file.endswith(".csv") and not file.startswith("__MACOSX") and "._" not in file:
            z.extract(file, path=f"/dbfs{extract_path}")
```

## 📂 1.2. Verificar arquivos extraídos

```
# Listar os arquivos extraídos
print("Arquivos extraídos:")
print(os.listdir(f"/dbfs{extract_path}"))
```
